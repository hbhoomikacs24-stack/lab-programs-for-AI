```python
import math


def alpha_beta_search(values, depth):
  best_score = -math.inf
  best_action = -1
  alpha = -math.inf
  beta = math.inf

  print("\n--- Tracing Alpha-Beta Search ---")
  for action in range(2):
    branch = "Left (0)" if action == 0 else "Right (1)"
    print(f"\nExploring Root -> {branch} branch:")
    v = min_value(1, action, values, depth, alpha, beta)
    if v > best_score:
      best_score = v
      best_action = action
    alpha = max(alpha, best_score)
    print(
        f"Root updated: best_score = {best_score}, alpha = {alpha}, beta ="
        f" {beta}"
    )

  return best_score, best_action


def max_value(current_depth, node_index, values, max_depth, alpha, beta):
  indent = "  " * current_depth
  if current_depth == max_depth:
    print(f"{indent}[Leaf] Value = {values[node_index]}")
    return values[node_index]

  v = -math.inf
  for i in range(2):
    child_index = node_index * 2 + i
    v = max(
        v,
        min_value(
            current_depth + 1, child_index, values, max_depth, alpha, beta
        ),
    )
    if v >= beta:
      print(
          f"{indent}[MAX Depth {current_depth}] Beta Cutoff! (v={v} >="
          f" beta={beta}) -> Pruning remaining branch"
      )
      return v
    alpha = max(alpha, v)
  return v


def min_value(current_depth, node_index, values, max_depth, alpha, beta):
  indent = "  " * current_depth
  if current_depth == max_depth:
    print(f"{indent}[Leaf] Value = {values[node_index]}")
    return values[node_index]

  v = math.inf
  for i in range(2):
    child_index = node_index * 2 + i
    v = min(
        v,
        max_value(
            current_depth + 1, child_index, values, max_depth, alpha, beta
        ),
    )
    if v <= alpha:
      print(
          f"{indent}[MIN Depth {current_depth}] Alpha Cutoff! (v={v} <="
          f" alpha={alpha}) -> Pruning remaining branch"
      )
      return v
    beta = min(beta, v)
  return v


# --- Take Input From User ---
depth = int(input("Enter the depth of the game tree (e.g., 3): "))
num_leaves = 2**depth

print(f"Enter {num_leaves} leaf node values separated by spaces:")
values = list(map(int, input().split()))

if len(values) != num_leaves:
  print(f"Error: You must enter exactly {num_leaves} numbers!")
else:
  optimal_val, action = alpha_beta_search(values, depth)
  branch_name = (
      "Left Branch (Action 0)" if action == 0 else "Right Branch (Action 1)"
  )
  print("\n--- Final Result ---")
  print(f"Optimal Value: {optimal_val}")
  print(f"Best Action from Root: {branch_name}")
```
